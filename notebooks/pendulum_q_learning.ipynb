## 2. Configuration

```python
import re
import random
import base64
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import imageio.v2 as imageio

import gymnasium as gym

from IPython import display as ipythondisplay
SEED = 42
random.seed(SEED)
np.random.seed(SEED)

ROOT = Path("as2_pendulum")
MODEL_DIR = ROOT / "models"
CHECKPOINT_DIR = ROOT / "checkpoints"
LOG_DIR = ROOT / "logs"
VIDEO_DIR = ROOT / "videos"

for folder in [
    MODEL_DIR,
    CHECKPOINT_DIR,
    LOG_DIR,
    VIDEO_DIR,
]:
    folder.mkdir(
        parents=True,
        exist_ok=True,
    )
```

## 3. Fixed Hyperparameters

```python
EPISODES = 30_000
EVAL_EPISODES = 200
CHECKPOINT_INTERVAL_EPISODES = 500
LOG_INTERVAL_EPISODES = 200
```

```python
N_COS = 15
N_SIN = 15
N_VEL = 25
N_ACTIONS = 11

ALPHA = 0.12
GAMMA = 0.99
EPSILON_START = 1.0
EPSILON_END = 0.05
EPSILON_DECAY = 0.995
```

## 4. Problem Formulation and Discretisation

**State:** `cos(θ)`, `sin(θ)`, angular velocity (15 × 15 × 25 bins).  
**Action:** Torque −2 to +2 (11 discrete choices).  
**Reward:** Negative angle, velocity and torque costs.  
**Goal:** Keep the pendulum upright with minimal effort.  
**Method:** Tabular Q-learning with ε-greedy exploration.

```python
ENV_ID = "Pendulum-v1"

cos_edges = np.linspace(
    -1.0,
    1.0,
    N_COS - 1,
)

sin_edges = np.linspace(
    -1.0,
    1.0,
    N_SIN - 1,
)

vel_edges = np.linspace(
    -8.0,
    8.0,
    N_VEL - 1,
)

actions = np.linspace(
    -2.0,
    2.0,
    N_ACTIONS,
    dtype=np.float32,
)


def discretise(
    observation,
):
    c, s, v = observation

    return (
        int(
            np.digitize(
                np.clip(
                    c,
                    -1,
                    1,
                ),
                cos_edges,
            )
        ),
        int(
            np.digitize(
                np.clip(
                    s,
                    -1,
                    1,
                ),
                sin_edges,
            )
        ),
        int(
            np.digitize(
                np.clip(
                    v,
                    -8,
                    8,
                ),
                vel_edges,
            )
        ),
    )
```

## 5. Utilities

```python
Q_LOG_PATH = (
    LOG_DIR
    / "q_learning_training_log.csv"
)


def latest_q_checkpoint():
    files = list(
        CHECKPOINT_DIR.glob(
            "pendulum_q_"
            "*_episodes.npz"
        )
    )

    if not files:
        return None, 0

    def get_episode(path):
        match = re.search(
            r"_(\d+)_episodes\.npz$",
            path.name,
        )

        return (
            int(match.group(1))
            if match
            else -1
        )

    latest = max(
        files,
        key=get_episode,
    )

    return (
        latest,
        get_episode(latest),
    )


def save_q_checkpoint(
    q_table,
    epsilon,
    completed_episodes,
):
    path = (
        CHECKPOINT_DIR
        / (
            "pendulum_q_"
            f"{completed_episodes}_episodes.npz"
        )
    )

    np.savez_compressed(
        path,
        q_table=q_table,
        epsilon=np.array(
            [epsilon],
            dtype=np.float32,
        ),
        completed_episodes=np.array(
            [completed_episodes],
            dtype=np.int32,
        ),
    )

    print(
        "Checkpoint saved:",
        path,
    )


def append_log(
    episode,
    reward,
    epsilon,
):
    row = pd.DataFrame([{
        "episode": episode,
        "reward": reward,
        "epsilon": epsilon,
    }])

    if Q_LOG_PATH.exists():
        existing = pd.read_csv(
            Q_LOG_PATH
        )

        row = pd.concat(
            [
                existing,
                row,
            ],
            ignore_index=True,
        )

        row = (
            row
            .drop_duplicates(
                subset=["episode"],
                keep="last",
            )
            .sort_values(
                "episode"
            )
        )

    row.to_csv(
        Q_LOG_PATH,
        index=False,
    )


def load_log():
    if not Q_LOG_PATH.exists():
        return pd.DataFrame()

    return (
        pd.read_csv(
            Q_LOG_PATH
        )
        .drop_duplicates(
            subset=["episode"],
            keep="last",
        )
        .sort_values(
            "episode"
        )
        .reset_index(
            drop=True
        )
    )


def evaluate_q(
    q_table,
    episodes=EVAL_EPISODES,
    seed_start=SEED + 10_000,
):
    env = gym.make(ENV_ID)
    rewards = []
    lengths = []
    seeds = []

    try:
        for episode in range(episodes):
            episode_seed = seed_start + episode
            observation, _ = env.reset(seed=episode_seed)
            state = discretise(observation)
            terminated = False
            truncated = False
            total_reward = 0.0
            total_length = 0

            while not (terminated or truncated):
                action_index = int(np.argmax(q_table[state]))
                action = np.array([actions[action_index]], dtype=np.float32)
                observation, reward, terminated, truncated, _ = env.step(action)
                state = discretise(observation)
                total_reward += float(reward)
                total_length += 1

            rewards.append(total_reward)
            lengths.append(total_length)
            seeds.append(episode_seed)
    finally:
        env.close()

    best_index = int(np.argmax(rewards))
    return {
        "mean_reward": float(np.mean(rewards)),
        "std_reward": float(np.std(rewards)),
        "median_reward": float(np.median(rewards)),
        "mean_length": float(np.mean(lengths)),
        "max_reward": float(rewards[best_index]),
        "best_seed": int(seeds[best_index]),
    }


def evaluate_q_checkpoints(episodes=EVAL_EPISODES):
    checkpoints = sorted(
        CHECKPOINT_DIR.glob("pendulum_q_*_episodes.npz"),
        key=lambda path: int(re.search(r"_(\d+)_episodes\.npz$", path.name).group(1)),
    )
    if not checkpoints:
        raise RuntimeError("No Q-learning checkpoints were found.")

    rows = []
    for checkpoint in checkpoints:
        with np.load(checkpoint) as saved:
            candidate = saved["q_table"]
            episode = int(saved["completed_episodes"][0])
        metrics = evaluate_q(candidate, episodes=episodes)
        rows.append({"episode": episode, **metrics})
        print(f"Evaluated Pendulum checkpoint: {episode:,} episodes")

    return pd.DataFrame(rows)

def random_baseline():
    env = gym.make(
        ENV_ID
    )

    rewards = []
    lengths = []

    for episode in range(
        EVAL_EPISODES
    ):
        observation, _ = env.reset(
            seed=SEED + 10_000 + episode
        )
        env.action_space.seed(SEED + 10_000 + episode)

        total_reward = 0.0
        total_length = 0
        terminated = False
        truncated = False

        while not (
            terminated
            or truncated
        ):
            observation, reward, terminated, truncated, _ = (
                env.step(
                    env.action_space.sample()
                )
            )

            total_reward += float(
                reward
            )

            total_length += 1

        rewards.append(
            total_reward
        )

        lengths.append(
            total_length
        )

    env.close()

    return {
        "mean_reward": float(
            np.mean(rewards)
        ),
        "std_reward": float(
            np.std(rewards)
        ),
        "median_reward": float(
            np.median(rewards)
        ),
        "mean_length": float(
            np.mean(lengths)
        ),
        "max_reward": float(
            np.max(rewards)
        ),
    }


def show_video(
    path,
):
    path = Path(path)

    video_b64 = (
        base64.b64encode(
            path.read_bytes()
        )
        .decode("ascii")
    )

    ipythondisplay.display(
        ipythondisplay.HTML(
            f"""
            <video controls
                   style="height: 400px;
                          max-width: 100%;">
                <source
                    src="data:video/mp4;base64,{video_b64}"
                    type="video/mp4">
            </video>
            """
        )
    )


def record_q_video(
    q_table,
    filename,
    seed=SEED + 5000,
):
    env = gym.make(
        ENV_ID,
        render_mode="rgb_array",
    )

    observation, _ = env.reset(
        seed=seed
    )

    state = discretise(
        observation
    )

    frames = [
        env.render()
    ]

    terminated = False
    truncated = False

    while not (
        terminated
        or truncated
    ):
        action_index = int(
            np.argmax(
                q_table[state]
            )
        )

        action = np.array(
            [
                actions[
                    action_index
                ]
            ],
            dtype=np.float32,
        )

        observation, reward, terminated, truncated, _ = (
            env.step(
                action
            )
        )

        state = discretise(
            observation
        )

        frames.append(
            env.render()
        )

    env.close()

    path = (
        VIDEO_DIR
        / filename
    )

    imageio.mimsave(
        path,
        frames,
        fps=30,
    )

    return path
```

## 6. Random Baseline

```python
random_result = (
    random_baseline()
)

random_result
```

## 7. Q-Learning Training

```python
checkpoint, completed = (
    latest_q_checkpoint()
)

if checkpoint is None:
    q_table = np.zeros(
        (
            N_COS,
            N_SIN,
            N_VEL,
            N_ACTIONS,
        ),
        dtype=np.float32,
    )

    epsilon = EPSILON_START
    completed = 0

else:
    saved = np.load(
        checkpoint
    )

    q_table = saved[
        "q_table"
    ]

    epsilon = float(
        saved["epsilon"][0]
    )

    completed = int(
        saved[
            "completed_episodes"
        ][0]
    )

    print(
        "Resuming from:",
        checkpoint,
    )

env = gym.make(
    ENV_ID
)

for episode in range(
    completed,
    EPISODES,
):
    observation, _ = env.reset(
        seed=SEED + episode
    )

    state = discretise(
        observation
    )

    total_reward = 0.0
    terminated = False
    truncated = False

    while not (
        terminated
        or truncated
    ):
        if (
            np.random.random()
            < epsilon
        ):
            action_index = (
                np.random.randint(
                    N_ACTIONS
                )
            )

        else:
            action_index = int(
                np.argmax(
                    q_table[state]
                )
            )

        action = np.array(
            [
                actions[
                    action_index
                ]
            ],
            dtype=np.float32,
        )

        next_observation, reward, terminated, truncated, _ = (
            env.step(
                action
            )
        )

        next_state = discretise(
            next_observation
        )

        bootstrap = (
            0.0
            if terminated
            else (
                GAMMA
                * np.max(
                    q_table[
                        next_state
                    ]
                )
            )
        )

        target = (
            float(reward)
            + bootstrap
        )

        index = (
            state
            + (
                action_index,
            )
        )

        q_table[index] += (
            ALPHA
            * (
                target
                - q_table[index]
            )
        )

        state = next_state

        total_reward += float(
            reward
        )

    completed_episodes = (
        episode + 1
    )

    append_log(
        completed_episodes,
        total_reward,
        epsilon,
    )

    epsilon = max(
        EPSILON_END,
        epsilon
        * EPSILON_DECAY,
    )

    if (
        completed_episodes
        % LOG_INTERVAL_EPISODES
        == 0
    ):
        recent = (
            load_log()
            .tail(
                LOG_INTERVAL_EPISODES
            )
        )

        print(
            f"episode="
            f"{completed_episodes:,} | "
            f"mean_reward="
            f"{recent['reward'].mean():.2f} | "
            f"epsilon="
            f"{epsilon:.3f}"
        )

    if (
        completed_episodes
        % CHECKPOINT_INTERVAL_EPISODES
        == 0
    ):
        save_q_checkpoint(
            q_table,
            epsilon,
            completed_episodes,
        )

env.close()

np.save(
    MODEL_DIR
    / "pendulum_q_table.npy",
    q_table,
)
```

## 8. Learning Curve

```python
q_curve = load_log()

q_curve["mean_100"] = (
    q_curve["reward"]
    .rolling(
        100,
        min_periods=1,
    )
    .mean()
)

plt.figure(
    figsize=(9, 4)
)

plt.plot(
    q_curve["episode"],
    q_curve["reward"],
    alpha=0.30,
    label="Episode reward",
)

plt.plot(
    q_curve["episode"],
    q_curve["mean_100"],
    linewidth=2,
    label="100-episode mean",
)

plt.title(
    "Pendulum Q-Learning Curve"
)
plt.xlabel("Episode")
plt.ylabel("Reward")
plt.legend()
plt.grid(alpha=0.25)
plt.show()

display(
    q_curve.tail(10)
)
```

## 9. Checkpoint Evaluation

Evaluate 100 seeded episodes per checkpoint and select the highest mean reward.

```python
q_checkpoint_results = evaluate_q_checkpoints(
    episodes=EVAL_EPISODES,
)

display(q_checkpoint_results)

best_q = (
    q_checkpoint_results
    .sort_values(
        ["mean_reward", "std_reward"],
        ascending=[False, True],
    )
    .iloc[0]
)

best_q_episode = int(best_q["episode"])
best_q_seed = int(best_q["best_seed"])
best_q_path = (
    CHECKPOINT_DIR
    / f"pendulum_q_{best_q_episode}_episodes.npz"
)

with np.load(best_q_path) as saved:
    best_q_table = saved["q_table"].copy()

q_result = {
    key: float(best_q[key])
    for key in (
        "mean_reward",
        "std_reward",
        "median_reward",
        "mean_length",
        "max_reward",
    )
}

print("Selected checkpoint episode:", best_q_episode)
print("Selected episode seed:", best_q_seed)
print("Selected mean reward:", q_result["mean_reward"])
```
