## 5. Environment
Breakout uses stacked image observations and discrete actions. The objective is to maximise cumulative reward.

```python
ENV_ID = "ale_py:ALE/Breakout-v5"


@contextlib.contextmanager
def suppress_stderr():
    try:
        stderr_fd = sys.stderr.fileno()
    except (AttributeError, io.UnsupportedOperation):
        yield
        return

    saved_fd = os.dup(stderr_fd)

    try:
        with open(os.devnull, "w") as devnull:
            os.dup2(
                devnull.fileno(),
                stderr_fd,
            )
            yield
    finally:
        os.dup2(
            saved_fd,
            stderr_fd,
        )
        os.close(saved_fd)


def make_breakout_env(
    seed,
    render_mode=None,
):
    def _init():
        with suppress_stderr():
            env = gym.make(
                ENV_ID,
                frameskip=1,
                repeat_action_probability=0.0,
                render_mode=render_mode,
            )

        env = AtariWrapper(
            env,
            screen_size=84,
            frame_skip=4,
            terminal_on_life_loss=False,
            clip_reward=False,
        )

        env = Monitor(env)
        env.reset(seed=seed)
        return env

    return _init


def build_env(
    seed=SEED,
    n_envs=1,
    render_mode=None,
):
    env = DummyVecEnv([
        make_breakout_env(
            seed + i,
            render_mode=render_mode,
        )
        for i in range(n_envs)
    ])

    return VecFrameStack(
        env,
        n_stack=4,
    )


print("Environment:", ENV_ID)
```

## 6. Custom CNN
A custom CNN extracts visual features from stacked Atari observations.

```python
class CustomCNN(BaseFeaturesExtractor):
    def __init__(
        self,
        observation_space,
        features_dim=CNN_FEATURES_DIM,
    ):
        super().__init__(
            observation_space,
            features_dim,
        )

        channels = observation_space.shape[0]

        self.cnn = nn.Sequential(
            nn.Conv2d(
                channels,
                32,
                kernel_size=8,
                stride=4,
            ),
            nn.ReLU(),
            nn.Conv2d(
                32,
                64,
                kernel_size=4,
                stride=2,
            ),
            nn.ReLU(),
            nn.Conv2d(
                64,
                64,
                kernel_size=3,
                stride=1,
            ),
            nn.ReLU(),
            nn.Flatten(),
        )

        with th.no_grad():
            sample = th.as_tensor(
                observation_space.sample()[None]
            ).float()

            n_flatten = self.cnn(
                sample
            ).shape[1]

        self.linear = nn.Sequential(
            nn.Linear(
                n_flatten,
                features_dim,
            ),
            nn.ReLU(),
        )

    def forward(
        self,
        observations,
    ):
        return self.linear(
            self.cnn(observations)
        )


POLICY_KWARGS = {
    "features_extractor_class": CustomCNN,
    "features_extractor_kwargs": {
        "features_dim": CNN_FEATURES_DIM,
    },
}
```

## 7. Utilities
Training logs, checkpoints, evaluation, and plotting are handled by reusable utilities.

```python
class TrainingLogCallback(BaseCallback):
    def __init__(
        self,
        log_path,
        algorithm,
        log_interval=100_000,
    ):
        super().__init__()

        self.log_path = Path(log_path)
        self.algorithm = algorithm
        self.log_interval = int(
            log_interval
        )

        self.next_log = None
        self.interval_start = None
        self.recent_rewards = []
        self.recent_lengths = []

        self.log_path.parent.mkdir(
            parents=True,
            exist_ok=True,
        )

    def _on_training_start(self):
        current = int(
            self.model.num_timesteps
        )

        self.interval_start = (
            current
            // self.log_interval
        ) * self.log_interval

        self.next_log = (
            self.interval_start
            + self.log_interval
        )

    def _write_row(
        self,
        row,
    ):
        new_row = pd.DataFrame(
            [row]
        )

        if self.log_path.exists():
            existing = pd.read_csv(
                self.log_path
            )

            new_row = pd.concat(
                [
                    existing,
                    new_row,
                ],
                ignore_index=True,
            )

            new_row = (
                new_row
                .drop_duplicates(
                    subset=["timestep"],
                    keep="last",
                )
                .sort_values(
                    "timestep"
                )
            )

        new_row.to_csv(
            self.log_path,
            index=False,
        )

    def _on_step(self):
        for info in self.locals.get(
            "infos",
            [],
        ):
            if "episode" in info:
                self.recent_rewards.append(
                    float(
                        info["episode"]["r"]
                    )
                )

                self.recent_lengths.append(
                    int(
                        info["episode"]["l"]
                    )
                )

        current = int(
            self.model.num_timesteps
        )

        if current < self.next_log:
            return True

        if self.recent_rewards:
            episodes = len(
                self.recent_rewards
            )

            mean_reward = float(
                np.mean(
                    self.recent_rewards
                )
            )

            mean_length = float(
                np.mean(
                    self.recent_lengths
                )
            )

        else:
            episodes = 0
            mean_reward = np.nan
            mean_length = np.nan

        row = {
            "algorithm": self.algorithm,
            "timestep": current,
            "interval_start": self.interval_start,
            "interval_end": self.next_log,
            "episodes": episodes,
            "ep_rew_mean": mean_reward,
            "ep_len_mean": mean_length,
        }

        self._write_row(row)

        line = (
            f"timestep={current:,} | "
            f"interval="
            f"{self.interval_start:,}-"
            f"{self.next_log:,} | "
            f"episodes={episodes} | "
            f"ep_rew_mean="
            f"{mean_reward:.2f} | "
            f"ep_len_mean="
            f"{mean_length:.1f}"
        )

        print(line)

        self.recent_rewards.clear()
        self.recent_lengths.clear()

        self.interval_start = (
            self.next_log
        )

        while (
            self.next_log
            <= current
        ):
            self.next_log += (
                self.log_interval
            )

        return True


class SaveOnIntervalCallback(BaseCallback):
    def __init__(
        self,
        save_interval,
        save_path,
        prefix,
        save_replay_buffer=False,
    ):
        super().__init__()

        self.save_interval = int(
            save_interval
        )

        self.save_path = Path(
            save_path
        )

        self.prefix = prefix
        self.save_replay_buffer = (
            save_replay_buffer
        )

        self.next_save = None

        self.save_path.mkdir(
            parents=True,
            exist_ok=True,
        )

    def _on_training_start(self):
        current = int(
            self.model.num_timesteps
        )

        self.next_save = (
            (
                current
                // self.save_interval
            )
            + 1
        ) * self.save_interval

    def _save_checkpoint(self):
        steps = int(
            self.model.num_timesteps
        )

        model_path = (
            self.save_path
            / (
                f"{self.prefix}_"
                f"{steps}_steps"
            )
        )

        self.model.save(
            model_path
        )

        if (
            self.save_replay_buffer
            and hasattr(
                self.model,
                "save_replay_buffer",
            )
        ):
            replay_path = (
                self.save_path
                / (
                    f"{self.prefix}_"
                    "latest_replay_buffer.pkl"
                )
            )

            self.model.save_replay_buffer(
                replay_path
            )

            step_path = (
                self.save_path
                / (
                    f"{self.prefix}_"
                    "latest_replay_buffer_steps.txt"
                )
            )

            step_path.write_text(
                str(steps),
                encoding="utf-8",
            )

        print(
            "Checkpoint saved:",
            str(model_path) + ".zip",
        )

    def _on_step(self):
        current = int(
            self.model.num_timesteps
        )

        if current >= self.next_save:
            self._save_checkpoint()

            while (
                self.next_save
                <= current
            ):
                self.next_save += (
                    self.save_interval
                )

        return True


def latest_step_checkpoint(
    folder,
    prefix,
):
    files = list(
        Path(folder).glob(
            f"{prefix}_*_steps.zip"
        )
    )

    if not files:
        return None, 0

    def get_steps(path):
        match = re.search(
            r"_(\d+)_steps\.zip$",
            path.name,
        )

        return (
            int(match.group(1))
            if match
            else -1
        )

    latest = max(
        files,
        key=get_steps,
    )

    return (
        latest,
        get_steps(latest),
    )


def load_training_log(
    path,
):
    path = Path(path)

    if not path.exists():
        return pd.DataFrame()

    return (
        pd.read_csv(path)
        .drop_duplicates(
            subset=["timestep"],
            keep="last",
        )
        .sort_values(
            "timestep"
        )
        .reset_index(
            drop=True
        )
    )


def plot_learning_curve(
    log_path,
    title,
    smoothing_points=5,
):
    frame = load_training_log(
        log_path
    )

    if frame.empty:
        print(
            "No saved training log found."
        )
        return frame

    frame["smoothed_reward"] = (
        frame["ep_rew_mean"]
        .rolling(
            smoothing_points,
            min_periods=1,
        )
        .mean()
    )

    plt.figure(
        figsize=(10, 5)
    )

    plt.plot(
        frame["timestep"],
        frame["ep_rew_mean"],
        alpha=0.35,
        label="100k interval mean",
    )

    plt.plot(
        frame["timestep"],
        frame["smoothed_reward"],
        linewidth=2,
        label=(
            f"{smoothing_points}-point "
            "moving mean"
        ),
    )

    plt.title(title)
    plt.xlabel("Timesteps")
    plt.ylabel(
        "Mean Episode Reward"
    )
    plt.legend()
    plt.grid(alpha=0.25)
    plt.show()

    return frame


def evaluate_checkpoint(
    model_class,
    checkpoint,
    episodes=EVAL_EPISODES,
    custom_objects=None,
    seed_start=SEED + 1000,
):
    model = model_class.load(
        checkpoint,
        device=DEVICE,
        custom_objects=custom_objects,
    )

    rewards = []
    lengths = []
    seeds = []

    for episode in range(episodes):
        episode_seed = (
            seed_start + episode
        )

        env = build_env(
            episode_seed,
            n_envs=1,
        )

        obs = env.reset()

        done = False
        total_reward = 0.0
        total_length = 0

        while not done:
            action, _ = model.predict(
                obs,
                deterministic=True,
            )

            obs, reward, done_array, _ = (
                env.step(action)
            )

            total_reward += float(
                reward[0]
            )

            total_length += 1

            done = bool(
                done_array[0]
            )

        env.close()

        rewards.append(
            total_reward
        )

        lengths.append(
            total_length
        )

        seeds.append(
            episode_seed
        )

    best_index = int(
        np.argmax(
            rewards
        )
    )

    result = {
        "mean_reward": float(
            np.mean(rewards)
        ),
        "std_reward": float(
            np.std(rewards)
        ),
        "median_reward": float(
            np.median(rewards)
        ),
        "mean_length": float(
            np.mean(lengths)
        ),
        "max_reward": float(
            rewards[best_index]
        ),
        "best_seed": int(
            seeds[best_index]
        ),
    }

    del model

    gc.collect()

    if th.cuda.is_available():
        th.cuda.empty_cache()

    return result
def evaluate_all_checkpoints(
    model_class,
    folder,
    prefix,
    custom_objects=None,
):
    files = list(
        Path(folder).glob(
            f"{prefix}_*_steps.zip"
        )
    )

    if not files:
        return pd.DataFrame()

    def get_steps(path):
        match = re.search(
            r"_(\d+)_steps\.zip$",
            path.name,
        )

        return int(
            match.group(1)
        )

    files = sorted(
        files,
        key=get_steps,
    )

    rows = []

    for checkpoint in files:
        steps = get_steps(
            checkpoint
        )

        print(
            f"Evaluating "
            f"{steps:,} steps..."
        )

        result = evaluate_checkpoint(
            model_class,
            checkpoint,
            custom_objects=custom_objects,
        )

        rows.append({
            "timestep": steps,
            **result,
        })

    return pd.DataFrame(
        rows
    )


def random_baseline(
    episodes=EVAL_EPISODES,
):
    env = build_env(
        SEED + 99,
        n_envs=1,
    )

    rewards = []
    lengths = []

    obs = env.reset()
    total_reward = 0.0
    total_length = 0

    while len(rewards) < episodes:
        action = np.array([
            env.action_space.sample()
        ])

        obs, reward, done, info = (
            env.step(action)
        )

        total_reward += float(
            reward[0]
        )

        total_length += 1

        if bool(done[0]):
            rewards.append(
                total_reward
            )

            lengths.append(
                total_length
            )

            total_reward = 0.0
            total_length = 0

    env.close()

    return {
        "mean_reward": float(
            np.mean(rewards)
        ),
        "std_reward": float(
            np.std(rewards)
        ),
        "median_reward": float(
            np.median(rewards)
        ),
        "mean_length": float(
            np.mean(lengths)
        ),
        "max_reward": float(
            np.max(rewards)
        ),
    }
```

Evaluate a random policy before training.

```python
random_result = (
    random_baseline()
)

random_result
```

## 9. PPO Training
PPO resumes automatically from the latest saved checkpoint.

```python
ppo_env = build_env(
    SEED + 1,
    n_envs=PPO_N_ENVS,
)

ppo_ckpt, completed_steps = (
    latest_step_checkpoint(
        PPO_CHECKPOINT_DIR,
        "breakout_ppo",
    )
)

ppo_config = {
    "learning_rate": PPO_LEARNING_RATE,
    "n_steps": PPO_N_STEPS,
    "batch_size": PPO_BATCH_SIZE,
    "n_epochs": PPO_N_EPOCHS,
    "gamma": PPO_GAMMA,
    "gae_lambda": PPO_GAE_LAMBDA,
    "clip_range": PPO_CLIP_RANGE,
    "ent_coef": PPO_ENT_COEF,
    "vf_coef": PPO_VF_COEF,
    "max_grad_norm": PPO_MAX_GRAD_NORM,
}

if ppo_ckpt is None:
    print(
        "No PPO checkpoint found. "
        "Starting from 0."
    )

    ppo = PPO(
        "CnnPolicy",
        ppo_env,
        policy_kwargs=POLICY_KWARGS,
        verbose=0,
        seed=SEED,
        device=DEVICE,
        **ppo_config,
    )

else:
    print(
        "Resuming PPO from:",
        ppo_ckpt,
    )

    ppo = PPO.load(
        ppo_ckpt,
        env=ppo_env,
        device=DEVICE,
        custom_objects=ppo_config,
    )

log_callback = TrainingLogCallback(
    PPO_LOG_PATH,
    "PPO",
    LOG_INTERVAL_STEPS,
)

save_callback = SaveOnIntervalCallback(
    CHECKPOINT_INTERVAL_STEPS,
    PPO_CHECKPOINT_DIR,
    "breakout_ppo",
    save_replay_buffer=False,
)

remaining_steps = max(
    0,
    TOTAL_TIMESTEPS
    - int(
        ppo.num_timesteps
    ),
)

print(
    "PPO environments:",
    PPO_N_ENVS,
)
print(
    "Rollout size:",
    PPO_N_STEPS
    * PPO_N_ENVS,
)
print(
    "Batch size:",
    PPO_BATCH_SIZE,
)
print(
    "Target steps:",
    TOTAL_TIMESTEPS,
)
print(
    "Current steps:",
    ppo.num_timesteps,
)
print(
    "Remaining steps:",
    remaining_steps,
)

if remaining_steps > 0:
    start = time.time()

    ppo.learn(
        total_timesteps=remaining_steps,
        reset_num_timesteps=False,
        callback=CallbackList([
            log_callback,
            save_callback,
        ]),
        progress_bar=True,
    )

    print(
        "Training time:",
        round(
            (
                time.time()
                - start
            )
            / 60,
            1,
        ),
        "minutes",
    )

ppo.save(
    MODEL_DIR
    / "breakout_ppo_final"
)
```

## 11. Performance Evaluation

Each saved checkpoint is evaluated on the same deterministic episode seeds. The checkpoint with the highest mean reward is selected, and the seed that produced its maximum evaluation reward is retained for video recording.

```python
ppo_results = (
    evaluate_all_checkpoints(
        PPO,
        PPO_CHECKPOINT_DIR,
        "breakout_ppo",
    )
)

display(
    ppo_results
)

best_ppo = (
    ppo_results
    .sort_values(
        "mean_reward",
        ascending=False,
    )
    .iloc[0]
)

print(
    "BEST CHECKPOINT"
)
print(
    "Timestep:",
    f"{int(best_ppo['timestep']):,}",
)
print(
    "Mean reward:",
    round(
        best_ppo[
            "mean_reward"
        ],
        2,
    ),
)
print(
    "Std reward:",
    round(
        best_ppo[
            "std_reward"
        ],
        2,
    ),
)
print(
    "Median reward:",
    round(
        best_ppo[
            "median_reward"
        ],
        2,
    ),
)
print(
    "Mean length:",
    round(
        best_ppo[
            "mean_length"
        ],
        2,
    ),
)
print(
    "Max reward:",
    round(
        best_ppo[
            "max_reward"
        ],
        2,
    ),
)

print(
    "Best seed:",
    int(
        best_ppo[
            "best_seed"
        ]
    ),
)
```

## 13. Record and Play Video

The best PPO checkpoint and its highest-reward evaluation seed are reused directly from the performance evaluation. The recorded video reproduces the best evaluated episode without running another search.

```python
def record_best_episode(
    model,
    prefix,
    seed,
    video_length=30_000,
):
    env = build_env(
        seed=seed,
        n_envs=1,
        render_mode="rgb_array",
    )

    env = VecVideoRecorder(
        env,
        video_folder=str(
            VIDEO_DIR
        ),
        record_video_trigger=(
            lambda step: step == 0
        ),
        video_length=video_length,
        name_prefix=prefix,
    )

    obs = env.reset()

    done = False

    while not done:
        action, _ = model.predict(
            obs,
            deterministic=True,
        )

        obs, _, done_array, _ = (
            env.step(action)
        )

        done = bool(
            done_array[0]
        )

    env.close()


def show_video(
    video_path,
    prefix,
):
    videos = sorted(
        Path(video_path).glob(
            f"{prefix}*.mp4"
        ),
        key=lambda path: (
            path.stat().st_mtime
        ),
    )

    if not videos:
        print(
            "No video found for prefix:",
            prefix,
        )
        return

    mp4 = videos[-1]

    video_b64 = (
        base64.b64encode(
            mp4.read_bytes()
        )
        .decode("ascii")
    )

    ipythondisplay.display(
        ipythondisplay.HTML(
            data=f"""
            <video controls
                   style="height: 400px;
                          max-width: 100%;">
                <source
                    src="data:video/mp4;base64,{video_b64}"
                    type="video/mp4">
            </video>
            """
        )
    )
```

```python
best_steps = int(
    best_ppo["timestep"]
)

best_seed = int(
    best_ppo["best_seed"]
)

best_checkpoint = (
    PPO_CHECKPOINT_DIR
    / (
        "breakout_ppo_"
        f"{best_steps}_steps.zip"
    )
)

video_model = PPO.load(
    best_checkpoint,
    device=DEVICE,
)

video_prefix = (
    "breakout-ppo-best-"
    f"{best_steps}-"
    f"seed-{best_seed}"
)

print(
    "Recording checkpoint:",
    f"{best_steps:,}",
)

print(
    "Evaluation best seed:",
    best_seed,
)

print(
    "Evaluation max reward:",
    best_ppo[
        "max_reward"
    ],
)

record_best_episode(
    video_model,
    prefix=video_prefix,
    seed=best_seed,
)

show_video(
    VIDEO_DIR,
    prefix=video_prefix,
)

del video_model

gc.collect()

if th.cuda.is_available():
    th.cuda.empty_cache()
```
